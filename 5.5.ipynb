# ecFbridge_combination

```python
import sys
import os
import pandas as pd

sys.path.append(r'./MEResource/')
sys.path.append(r'./script/')
from ecFbridge_combination import IntegratedTargetIdentifierWithOptimization

def batch_predict(product_ids, model_file, base_output_dir='prediction_batch'):
    """串行批量预测 - 所有文件在同一文件夹"""
    os.makedirs(base_output_dir, exist_ok=True)
    
    results = []
    
    for idx, product_id in enumerate(product_ids, 1):
        print(f"\n[{idx}/{len(product_ids)}] Processing {product_id}...")
        
        try:
            # 创建识别器，输出直接到base_output_dir
            identifier = IntegratedTargetIdentifierWithOptimization(
                model_file=model_file,
                product_id=product_id,
                improvement_threshold=0.1,
                exclusion_threshold=-3,
                output_dir=base_output_dir,  # 直接使用主文件夹
                eta_threshold=1,
                enable_distance_filter=True,
                enable_ibridge=True,
                enable_combination_optimization=True,
                verbose_flux=False
            )
            
            final_targets = identifier.run_integrated_pipeline()
            
            # 记录结果
            results.append({
                'product_id': product_id,
                'status': 'SUCCESS',
                'OE_genes': len(final_targets.get('OE', [])),
                'KD_genes': len(final_targets.get('KD', [])),
            })
            
            print(f"✓ {product_id} completed: OE={len(final_targets.get('OE', []))}, KD={len(final_targets.get('KD', []))}")
            
        except Exception as e:
            results.append({
                'product_id': product_id,
                'status': 'FAILED',
                'OE_genes': 0,
                'KD_genes': 0,
            })
            print(f"✗ {product_id} failed: {e}")
    
    # 保存汇总结果
    df = pd.DataFrame(results)
    summary_file = os.path.join(base_output_dir, 'batch_summary.csv')
    df.to_csv(summary_file, index=False)
    
    print(f"\nComplete! Summary saved to: {summary_file}")
    print(f"Total files in {base_output_dir}: {len(os.listdir(base_output_dir))}")
    return df

# 使用示例
if __name__ == "__main__":
    product_ids = [
        "cpd00036"
    ]
    
    model_file = './model_Calibration/DLKcat_Calibration/ecCZ870_DLKcat_Calibration_modify.json'
    
    results = batch_predict(
        product_ids=product_ids,
        model_file=model_file,
        base_output_dir='prediction_target_ecFbridge_combination(0122)'
    )
```
