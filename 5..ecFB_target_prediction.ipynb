# Biomass Composition Sensitivity Analysis

```python
import sys
import warnings

warnings.filterwarnings("ignore")
sys.path.insert(0, "./ecscript/")

from batch_predict_parallel import batch_predict

PRODUCT_IDS = [
    "cpd00039", "cpd00322", "cpd00060", "cpd00107",
    "cpd00156", "cpd00161", "cpd00227", "cpd00129",
    "cpd00051", "cpd00064", "cpd00274", "cpd00069",
    "cpd00065", "cpd00054", "cpd00084", "cpd00119",
    "cpd00066",
]

CASES = {
     "Protein +10%": (
         "./model_Calibration/DLKcat_Calibration/ecCZ870_Pprotein1.1.json",
         "./ecfb_Pprotein1.1",
     ),
     "Protein -10%": (
         "./model_Calibration/DLKcat_Calibration/ecCZ870_Pprotein0.9.json",
         "./ecfb_Pprotein0.9",
     ),
     "Cell wall +10%": (
         "./model_Calibration/DLKcat_Calibration/ecCZ870_Pcellwall1.1.json",
         "./ecfb_Pcellwall1.1",
     ),
    "Cell wall -10%": (
         "./model_Calibration/DLKcat_Calibration/ecCZ870_Pcellwall0.9.json",
         "./ecfb_Pcellwall0.9",
    ),
     "Lipids -10%": (
          "./model_Calibration/DLKcat_Calibration/ecCZ870_lipids0.9.json",
          "./ecfb_lipids0.9",
     ),
     "Lipids +10%": (
          "./model_Calibration/DLKcat_Calibration/ecCZ870_lipids1.1.json",
          "./ecfb_lipids1.1",
    ),
}

COMMON_ARGS = {
    "product_ids": PRODUCT_IDS,
    "method": "ecfb",
    "biomass_rxn": "Bio_cgATCC14067_c",
    "c_source_rxn": "EX_cpd00027_e_reverse",
    "ecb_num_step": 20,
    "ecb_bio_fraction": 0.1,
    "ecb_threshold": 0.01,
    "ecb_top_score_percentage": 0.9,
    "efs_fseof_steps": 16,
    "efs_top_percentage": 0.9,
    "st_improvement_threshold": 0.1,
    "st_exclusion_threshold": -5.0,
    "n_workers": 4,
    "show_progress": True,
    "save_interval": 3,
    "debug_output": False,
    "naming": "modelseed",
}


for case_name, (model_file, output_dir) in CASES.items():
    print(f"\n{'=' * 60}")
    print(f"Running: {case_name}")
    print(f"{'=' * 60}")

    summary_df, saved_dir = batch_predict(
        model_file=model_file,
        output_base_dir=output_dir,
        **COMMON_ARGS,
    )

    print(summary_df.to_string(index=False))
    print(f"\nResults saved in: {saved_dir}")
```
